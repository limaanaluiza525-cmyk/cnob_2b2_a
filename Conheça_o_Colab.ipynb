##Fórmula da velocidade

```
def calculaVelMedia(deltaS, deltaT):
  velMedia = (deltaS/deltaT)
  return velMedia
```

##Teste Serviço Nuvem: Velocidade Média

```
print("Vm=", calculaVelMedia(30, 15))
```

##Fórmula da aceleração

```
def calculaAceleracao(vf, vi, t):
  return(vf - vi) / t
```
